```python
# Create a figure and axis object with size 8x5 inches
fig, ax = plt.subplots(figsize=(8,5))

# Sample data - monthly sales revenue
months = ['Jan','Feb','Mar','Apr','May','Jun']
sales = [42000,55000,49000,67000,71000,83000]

# Plot a line graph
# color = line color
# linewidth = thickness of line
# marker = circle points on each data value
# markersize = size of markers
ax.plot(months, sales, color="#008408", linewidth=2.5, marker='o', markersize=7)

# Set chart title
ax.set_title("Monthly Sales 2024", fontsize=14)

# Set X-axis label
ax.set_xlabel("Months", fontsize=12)

# Set Y-axis label
ax.set_ylabel("Revenue($)", fontsize=12)

# Add grid lines only on Y-axis
ax.grid(axis='y', alpha=0.4)

# Adjust layout to prevent overlapping
plt.tight_layout()

plt.show()
```

```python
# 📘 Scenario Question
# A company wants to analyze whether advertising spend impacts sales revenue. The marketing team collected data from
#  several campaigns, where ad spend ranged from $2,000 to $18,000 and the corresponding sales revenue was recorded.
# You are asked to:
# - Plot a scatter chart showing the relationship between ad spend and sales.
# - Add a trend line to highlight the overall pattern.
# - Interpret the chart: does higher ad spend generally lead to higher sales?

# Data: Ad spend V/S sales
import matplotlib.pyplot as plt
import numpy as np

# Data: Ad spend vs Sales
ad_spend = np.array([2000,4000,6000,8000,10000,12000,14000,16000,18000])
sales = np.array([7000,15000,21000,25000,32000,36000,42000,47000,52000])

# Scatter plot
plt.scatter(ad_spend, sales, color='blue', alpha=0.7)

# Trend Line (Linear Regression Line)
z = np.polyfit(ad_spend, sales, 1)   # degree 1 = straight line
p = np.poly1d(z)

plt.plot(ad_spend, p(ad_spend), 'r--', label='Trend Line')

# Labels and title
plt.xlabel('Ad Spend ($)')
plt.ylabel('Sales ($)')
plt.title('Ad Spend vs Sales')
plt.legend()

plt.show()
```

```python
# Probability  

import random
heads = 0
total = 1000
for i in range(total):
    result = random.randint(0,1)
    if result == 1:
        heads = heads + 1
tails = total - heads

print("Heads:", heads)
print("Tails:", tails)
print("P(Heads)=", heads/total)
```

```python
# Distribution

import numpy as np
import matplotlib.pyplot as plt

#200 random heights, avg=165, spread=8
heights=np.random.normal(165,8,200)

print("Shortest:",round(heights.min(),1))
print("Tallest:",round(heights.max(),1))
print("Average:",round(heights.mean(),1))

#Draw the bell shape
plt.hist(heights,bins=20,
color="#00BFA6",
edgecolor="white")

plt.axvline(heights.mean(),color="red",linewidth=2,label='Average')

plt.title("Height of 200 students")
plt.xlabel("height centimeters")
plt.ylabel("No of students")
plt.legend()
plt.show()
```
