```python
# 🎯 Example 1: Sequential Multi-Agent System (Pipeline)
# 🧠 Scenario

# “A travel company has different employees (agents):

# Planner → decides steps

# Flight Agent → finds flights

# Weather Agent → checks weather

# Decision Agent → gives final answer”

# AGENT 1: PLANNER

def planner_agent(user_query):
    print("\n[Planner Agent] Creating plan...")
    return ["flight", "weather", "decision"]


# AGENT 2: FLIGHT AGENT

def flight_agent():
    print("\n[Flight Agent] Fetching flights...")
    return [
        {"airline": "IndiGo", "price": 4500},
        {"airline": "Air India", "price": 5200}
    ]


# AGENT 3: WEATHER AGENT

def weather_agent():
    print("\n[Weather Agent] Checking weather...")
    return {"condition": "Clear", "temp": 28}


# AGENT 4: DECISION AGENT

def decision_agent(flights, weather):
    print("\n[Decision Agent] Making decision...")

    cheapest = min(flights, key=lambda x: x["price"])

    if weather["condition"] == "Rain":
        return "Avoid travel due to bad weather"

    return f"Book {cheapest['airline']} at ₹{cheapest['price']}"


# MAIN MULTI-AGENT SYSTEM

def travel_multi_agent(user_query):
    print("User Query:", user_query)

    plan = planner_agent(user_query)

    flights = None
    weather = None

    for step in plan:
        if step == "flight":
            flights = flight_agent()

        elif step == "weather":
            weather = weather_agent()

        elif step == "decision":
            result = decision_agent(flights, weather)

    return result


# RUN
response = travel_multi_agent("Plan my trip Delhi to Mumbai")
print("\nFinal Answer:", response)
```

```python
# Scenario
# “A hospital uses different employees (agents) to handle patient care in sequence.”

# ================================
# AGENT 1: Intake Agent (Planner)
# - Collects patient symptoms and history
# - Decides which steps are needed (tests, consultations, etc.)

# ================================
# AGENT 2: Diagnostic Agent
# - Orders lab tests or scans
# - Interprets results and identifies possible conditions

# ================================
# AGENT 3: Treatment Agent
# - Suggests treatment options (medication, therapy, surgery)
# - Considers patient preferences and medical guidelines

# ================================
# AGENT 4: Decision Agent
# - Reviews all inputs (history, diagnostics, treatment options)
# - Provides the final recommendation to the patient


# Scenario:
# A hospital uses multiple AI agents to handle patient care step by step.


# AGENT 1: Intake Agent (Planner)

def planner_agent(user_query):
    print("\n[Planner Agent] Collecting patient information...")
    print("Symptoms received:", user_query)

    # Decide which agents should run
    return ["diagnostic", "treatment", "decision"]


# AGENT 2: Diagnostic Agent

def diagnostic_agent():
    print("\n[Diagnostic Agent] Analyzing symptoms and test results...")

    diagnosis = [
        {"disease": "Skin Allergy"},
        {"disease": "Viral Infection"}
    ]

    return diagnosis


# AGENT 3: Treatment Agent

def treatment_agent():
    print("\n[Treatment Agent] Suggesting treatment options...")

    treatment = {
        "medicine": "Paracetamol",
        "rest_days": 3
    }

    return treatment

# AGENT 4: Decision Agent

def decision_agent(diagnostic, treatment):
    print("\n[Decision Agent] Reviewing diagnosis and treatment...")

    disease = diagnostic[0]["disease"]

    return f"Final Recommendation: Patient likely has {disease}. Prescribe {treatment['medicine']} and rest for {treatment['rest_days']} days."


# MAIN MULTI-AGENT SYSTEM

def hospital_multi_agent(user_query):

    print("Patient Query:", user_query)

    plan = planner_agent(user_query)

    diagnostic = None
    treatment = None

    for step in plan:

        if step == "diagnostic":
            diagnostic = diagnostic_agent()

        elif step == "treatment":
            treatment = treatment_agent()

        elif step == "decision":
            result = decision_agent(diagnostic, treatment)

    return result


# RUN SYSTEM
response = hospital_multi_agent("Patient reports itching and mild fever")
print("\nFinal Answer:", response)
```

```python

# AGENT 1: Crisis Coordinator (Broadcaster)


def crisis_coordinator():
    message = "Data breach detected in customer database. Immediate response required."
    print("\n[Crisis Coordinator] Broadcasting alert:")
    print(message)
    return message


# AGENT 2: IT Security Agent


def it_security_agent(message):
    print("\n[IT Security Agent] Received broadcast.")
    return "Isolate affected servers, patch vulnerabilities, start forensic analysis."

# AGENT 3: Communications Agent

def communications_agent(message):
    print("\n[Communications Agent] Received broadcast.")
    return "Draft internal memo, prepare press release, notify stakeholders."


# AGENT 4: Finance Agent


def finance_agent(message):
    print("\n[Finance Agent] Received broadcast.")
    return "Estimate financial impact, allocate emergency funds, review insurance coverage."

# AGENT 5: Legal Agent

def legal_agent(message):
    print("\n[Legal Agent] Received broadcast.")
    return "Assess regulatory obligations, prepare compliance reports, advise on liability."

# AGENT 6: HR Agent

def hr_agent(message):
    print("\n[HR Agent] Received broadcast.")
    return "Brief employees, provide guidance on handling customer queries, ensure morale support."

# AGENT 7: Decision Agent

def decision_agent(responses):

    print("\n[Decision Agent] Integrating responses...")

    final_plan = "\nFinal Crisis Response Plan:\n"

    for agent, response in responses.items():
        final_plan += f"{agent}: {response}\n"

    return final_plan

# MAIN MULTI-AGENT SYSTEM

def crisis_response_system():

    # Broadcast message
    alert = crisis_coordinator()

    responses = {}

    responses["IT Security"] = it_security_agent(alert)
    responses["Communications"] = communications_agent(alert)
    responses["Finance"] = finance_agent(alert)
    responses["Legal"] = legal_agent(alert)
    responses["HR"] = hr_agent(alert)

    # Final decision
    result = decision_agent(responses)

    return result


# RUN SYSTEM
response = crisis_response_system()

print(response)
```

```python
from groq import Groq
from dotenv import load_dotenv
import os

load_dotenv()

client = Groq(api_key=os.getenv("api_key"))

# Function to query the LLM
def query_llm(prompt):

    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[
            {"role": "user", "content": prompt}
        ]
    )

    return response.choices[0].message.content


# AGENT 1: Crisis Coordinator (Broadcaster)

def crisis_coordinator():

    message = "Data breach detected in customer database. Immediate response required."

    print("\n[Crisis Coordinator] Broadcasting alert:")
    print(message)

    return message


# AGENT 2: IT Security Agent

def it_security_agent(alert):

    print("\n[IT Security Agent] Responding...")

    prompt = f"""
    Crisis Alert: {alert}

    As an IT Security specialist, describe immediate technical actions.
    """

    return query_llm(prompt)


# AGENT 3: Communications Agent

def communications_agent(alert):

    print("\n[Communications Agent] Responding...")

    prompt = f"""
    Crisis Alert: {alert}

    As a corporate communications expert, describe communication actions.
    """

    return query_llm(prompt)


# AGENT 4: Finance Agent

def finance_agent(alert):

    print("\n[Finance Agent] Responding...")

    prompt = f"""
    Crisis Alert: {alert}

    As a finance manager, estimate financial impact and funding needs.
    """

    return query_llm(prompt)

# AGENT 5: Legal Agent

def legal_agent(alert):

    print("\n[Legal Agent] Responding...")

    prompt = f"""
    Crisis Alert: {alert}

    As a legal advisor, outline regulatory and compliance actions.
    """

    return query_llm(prompt)


# AGENT 6: HR Agent

def hr_agent(alert):

    print("\n[HR Agent] Responding...")

    prompt = f"""
    Crisis Alert: {alert}

    As an HR manager, describe how employees should be informed and supported.
    """

    return query_llm(prompt)


# AGENT 7: Decision Agent

def decision_agent(responses):

    print("\n[Decision Agent] Integrating responses...")

    prompt = f"""
    Integrate the following crisis responses into one unified action plan.

    {responses}

    Produce a final coordinated response plan.
    """

    return query_llm(prompt)


# MAIN SYSTEM

def crisis_response_system():

    alert = crisis_coordinator()

    responses = {}

    responses["IT Security"] = it_security_agent(alert)
    responses["Communications"] = communications_agent(alert)
    responses["Finance"] = finance_agent(alert)
    responses["Legal"] = legal_agent(alert)
    responses["HR"] = hr_agent(alert)

    final_plan = decision_agent(responses)

    return final_plan


# RUN SYSTEM
result = crisis_response_system()

print("\nFinal Crisis Response Plan:\n")
print(result)
```

```python
# Scenario: Corporate Product Launch Broadcast
# Imagine a company preparing to launch Product X in Q3. The Coordinator Agent (like a corporate program manager) sends out a broadcast announcement to all departments at once:
# “Product X launch in Q3, target market North America, budget $5M.”


# 📢 Coordinator Agent (Broadcaster)
# - Sends the launch announcement to all departments simultaneously.
# - This is implemented in the code by the broadcast() function, which uses asyncio.gather() to run all agents in parallel.

# 📈 Marketing Agent
# - Receives the broadcast.
# - Responds with a marketing strategy: campaigns, channels, and positioning.
# 💰 Finance Agent
# - Receives the broadcast.
# - Responds with budget allocation and ROI forecasts.
# 🏭 Operations Agent
# - Receives the broadcast.
# - Responds with production and supply chain actions.
# 👩‍⚖️ Legal Agent
# - Receives the broadcast.
# - Responds with compliance checks and contract actions.
# 👩‍💼 HR Agent
# - Receives the broadcast.
# - Responds with staffing and training plans.

# 🧑‍⚖️ Decision Agent
# - Collects all responses.
# - Integrates them into a Final Corporate Launch Plan.
# - In the code, this is the decision_agent() function that merges all outputs into one consolidated plan.
```

```python
from groq import Groq
from dotenv import load_dotenv
import os

load_dotenv()

client = Groq(api_key=os.getenv("api_key"))


# ================================
# LLM QUERY
# ================================
def query_llm(prompt):

    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[
            {"role": "user", "content": prompt}
        ]
    )

    return response.choices[0].message.content


# ================================
# AGENT 1 – Crisis Coordinator (Broadcast)
# ================================
def broadcaster_agent():

    message = (
        "Data breach detected in customer database. "
        "Immediate response required."
    )

    print("\n[Coordinator] Broadcasting crisis message...")

    return message


# ================================
# AGENT 2 – IT Security
# ================================
def it_agent(message):

    print("\n[IT Security Agent]")

    prompt = f"""
    Crisis message: {message}

    What IT security actions needed?
    """

    return query_llm(prompt)


# ================================
# AGENT 3 – Communications
# ================================
def communication_agent(message):

    print("\n[Communication Agent]")

    prompt = f"""
    Crisis message: {message}

    What communication steps needed?
    """

    return query_llm(prompt)


# ================================
# AGENT 4 – Finance
# ================================
def finance_agent(message):

    print("\n[Finance Agent]")

    prompt = f"""
    Crisis message: {message}

    What financial actions needed?
    """

    return query_llm(prompt)


# ================================
# AGENT 5 – Legal
# ================================
def legal_agent(message):

    print("\n[Legal Agent]")

    prompt = f"""
    Crisis message: {message}

    What legal steps required?
    """

    return query_llm(prompt)


# ================================
# AGENT 6 – HR
# ================================
def hr_agent(message):

    print("\n[HR Agent]")

    prompt = f"""
    Crisis message: {message}

    What HR actions needed?
    """

    return query_llm(prompt)


# ================================
# AGENT 7 – Decision Agent
# ================================
def decision_agent(responses):

    print("\n[Decision Agent] Creating final plan...")

    prompt = f"""
    IT: {responses['it']}

    Communication: {responses['comm']}

    Finance: {responses['finance']}

    Legal: {responses['legal']}

    HR: {responses['hr']}

    Create final crisis response plan.
    """

    return query_llm(prompt)


# ================================
# MAIN MULTI AGENT SYSTEM
# ================================
def crisis_multi_agent():

    message = broadcaster_agent()

    responses = {}

    # Broadcast to all agents
    responses["it"] = it_agent(message)
    responses["comm"] = communication_agent(message)
    responses["finance"] = finance_agent(message)
    responses["legal"] = legal_agent(message)
    responses["hr"] = hr_agent(message)

    final = decision_agent(responses)

    return final


# ================================
# RUN SYSTEM
# ================================
result = crisis_multi_agent()

print("\nFinal Crisis Plan:\n", result)
```

```python
import os
from dotenv import load_dotenv
from groq import Groq

# Load environment variables
load_dotenv()

client = Groq(api_key=os.getenv("api_key"))

# SHARED MEMORY

class SharedMemory:

    def __init__(self):
        self.storage = {}

    def update(self, agent, data):
        self.storage[agent] = data

    def get(self, agent):
        return self.storage.get(agent)

    def all(self):
        return self.storage


memory = SharedMemory()


# LLM CALL FUNCTION

def query_llm(prompt):

    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[{"role": "user", "content": prompt}]
    )

    return response.choices[0].message.content


# SEARCH AGENT

class SearchAgent:

    name = "search"

    def run(self, goal, memory):

        print("\n[Search Agent] Gathering EV industry information...")

        prompt = f"""
        Goal: {goal}

        Collect key EV market facts and trends in India for 2025.
        """

        result = query_llm(prompt)

        return result

# ANALYST AGENT

class AnalystAgent:

    name = "analyst"

    def run(self, goal, memory):

        print("\n[Analyst Agent] Interpreting data...")

        search_data = memory.get("search")

        prompt = f"""
        Analyze the following EV market data:

        {search_data}

        Provide key insights and trends.
        """

        return query_llm(prompt)

# WRITER AGENT

class WriterAgent:

    name = "writer"

    def run(self, goal, memory):

        print("\n[Writer Agent] Drafting report...")

        analysis = memory.get("analyst")

        prompt = f"""
        Using the analysis below, write a structured report.

        {analysis}

        Topic: EV industry trends in India for 2025.
        """

        return query_llm(prompt)


# QA AGENT

class QAAgent:

    name = "qa"

    def run(self, goal, memory):

        print("\n[QA Agent] Checking report quality...")

        draft = memory.get("writer")

        prompt = f"""
        Review this report for accuracy, clarity.

        {draft}

        Suggest improvements or corrections.
        """

        return query_llm(prompt)

# MAIN-MULTI AGENT SYSTEM

def orchestrator(goal):

    print("Goal:", goal)

    search = SearchAgent()
    analyst = AnalystAgent()
    writer = WriterAgent()
    qa = QAAgent()

    agents = [search, analyst, writer, qa]

    for agent in agents:

        result = agent.run(goal, memory)

        memory.update(agent.name, result)

    final_report = memory.get("writer")

    print("FINAL REPORT")

    print(final_report)

    print("QA FEEDBACK")


    print(memory.get("qa"))

# RUN PIPELINE

goal = "Write a comprehensive market report on EV industry trends in India for 2025."

orchestrator(goal)
```
